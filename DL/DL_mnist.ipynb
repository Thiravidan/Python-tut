```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
(Xtrain , ytrain) , (Xtest , ytest) = keras.datasets.mnist.load_data()
```

```python
Xtrain.shape
```

```python
Xtrain[0].shape
```

```python
plt.matshow(Xtrain[0])
```

```python
ytrain[0]
```

```python
Xtrain = Xtrain / 255
```

```python
Xtest = Xtest / 255
```

```python
Xtrain[0]
```

```python
Xtest[0]
```

```python
Xtrain_flattened = Xtrain.reshape(len(Xtrain), 28*28)
Xtest_flattened = Xtest.reshape(len(Xtest), 28*28)
```

```python
Xtest_flattened.shape
```

Simple Neural network without hidden layers

```python
model = keras.Sequential([
    keras.layers.Dense(10,input_shape=(784,),activation='sigmoid')
])

```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(Xtrain_flattened,ytrain,epochs=10)
```

```python
model.evaluate(Xtest_flattened,ytest)
```

```python
ypred = model.predict(Xtest_flattened)
ypred[0]
```

```python
plt.matshow(Xtest[1])
```

```python
np.argmax(ypred[1])
```

```python
ypred_labels = [np.argmax(i) for i in ypred]
```

```python
ypred_labels[:5]
```

```python
cm = tf.math.confusion_matrix(labels=ytest,predictions=ypred_labels)
cm
```

```python
import seaborn as sns
plt.figure(figsize=(10,7))
sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

Using a Hidden Layer

```python
model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(Xtrain_flattened,ytrain,epochs=5)
```

```python
model.evaluate(Xtest_flattened,ytest)
```

```python
y_predicted = model.predict(Xtest_flattened)
y_predicted_labels = [np.argmax(i) for i in y_predicted]
cm = tf.math.confusion_matrix(labels=ytest,predictions=y_predicted_labels)

plt.figure(figsize = (10,7))
sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

using flatten layer

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(100 , activation='relu'),
    keras.layers.Dense(10 , activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(Xtrain , ytrain , epochs=10)
```

```python
model.evaluate(Xtest,ytest)
```
